# IMPORTS

```python
import pandas as pd
import numpy as np
import os

pd.set_option("display.max_columns", None)
```

# File paths

```python
election_path = "../data/raw/countypres_2000-2024.csv"
model_ready_path = "../data/processed/model_ready_dataset.csv"

output_path = "../data/processed/historical_model_ready_dataset.csv"
```

# Load datasets

```python
election_df = pd.read_csv(election_path)

model_df = pd.read_csv(
    model_ready_path,
    dtype={"county_fips": str},
    low_memory=False
)

model_df["county_fips"] = model_df["county_fips"].astype(str).str.zfill(5)

print("Election dataset shape:", election_df.shape)
print("Model-ready dataset shape:", model_df.shape)

display(election_df.head())
display(model_df.head())
```

# Check available election years

```python
print("Available election years:")
print(sorted(election_df["year"].dropna().unique()))
```

# Clean election FIPS and filter parties/years

```python
valid_years = [2000, 2004, 2008, 2012, 2016, 2020, 2024]

election_clean = election_df[
    election_df["year"].isin(valid_years)
].copy()

election_clean = election_clean[
    election_clean["party"].isin(["DEMOCRAT", "REPUBLICAN"])
].copy()

election_clean["county_fips"] = pd.to_numeric(
    election_clean["county_fips"],
    errors="coerce"
)

election_clean = election_clean[
    election_clean["county_fips"].notna()
].copy()

election_clean["county_fips"] = (
    election_clean["county_fips"]
    .astype(int)
    .astype(str)
    .str.zfill(5)
)

print(election_clean.shape)
display(election_clean.head())
```

# Pivot election data to county-year level

```python
election_pivot = election_clean.pivot_table(
    index=["county_fips", "year", "state", "county_name", "totalvotes"],
    columns="party",
    values="candidatevotes",
    aggfunc="sum"
).reset_index()

election_pivot.columns.name = None

election_pivot = election_pivot.rename(columns={
    "DEMOCRAT": "democrat_votes",
    "REPUBLICAN": "republican_votes"
})

election_pivot["democrat_votes"] = election_pivot["democrat_votes"].fillna(0)
election_pivot["republican_votes"] = election_pivot["republican_votes"].fillna(0)

election_pivot["dem_vote_share"] = (
    election_pivot["democrat_votes"] / election_pivot["totalvotes"]
)

election_pivot["party_winner"] = np.where(
    election_pivot["democrat_votes"] > election_pivot["republican_votes"],
    1,
    0
)

print(election_pivot.shape)
display(election_pivot.head())
```

# Check county/year coverage

```python
coverage = election_pivot.groupby("year")["county_fips"].nunique()

print("County count per year:")
display(coverage)
```

# Create wide historical vote-share table

```python
dem_share_wide = election_pivot.pivot_table(
    index="county_fips",
    columns="year",
    values="dem_vote_share",
    aggfunc="first"
)

dem_share_wide.columns = [
    f"dem_share_{int(col)}" for col in dem_share_wide.columns
]

dem_share_wide = dem_share_wide.reset_index()

print(dem_share_wide.shape)
display(dem_share_wide.head())
```

# Create wide party-winner table

```python
winner_wide = election_pivot.pivot_table(
    index="county_fips",
    columns="year",
    values="party_winner",
    aggfunc="first"
)

winner_wide.columns = [
    f"winner_{int(col)}" for col in winner_wide.columns
]

winner_wide = winner_wide.reset_index()

print(winner_wide.shape)
display(winner_wide.head())
```

# Merge historical election tables

```python
historical_features = dem_share_wide.merge(
    winner_wide,
    on="county_fips",
    how="inner"
)

print(historical_features.shape)
display(historical_features.head())
```

# Create historical features from 2000–2020 only

```python
train_years = [2000, 2004, 2008, 2012, 2016, 2020]

dem_share_cols = [f"dem_share_{year}" for year in train_years]
winner_cols = [f"winner_{year}" for year in train_years]

historical_features["avg_dem_share_2000_2020"] = historical_features[dem_share_cols].mean(axis=1)

historical_features["dem_wins_count_2000_2020"] = historical_features[winner_cols].sum(axis=1)

historical_features["vote_volatility_2000_2020"] = historical_features[dem_share_cols].std(axis=1)

historical_features["swing_2016_2020"] = (
    historical_features["dem_share_2020"] -
    historical_features["dem_share_2016"]
)

historical_features["swing_2012_2020"] = (
    historical_features["dem_share_2020"] -
    historical_features["dem_share_2012"]
)

historical_features["swing_2000_2020"] = (
    historical_features["dem_share_2020"] -
    historical_features["dem_share_2000"]
)
```

# Create 2024 target variables

```python
historical_features["target_2024_party_winner"] = historical_features["winner_2024"]

historical_features["target_2024_dem_vote_share"] = historical_features["dem_share_2024"]

display(
    historical_features[
        [
            "county_fips",
            "avg_dem_share_2000_2020",
            "dem_wins_count_2000_2020",
            "vote_volatility_2000_2020",
            "swing_2016_2020",
            "target_2024_party_winner",
            "target_2024_dem_vote_share"
        ]
    ].head()
)
```

# Check missing values in historical features

```python
historical_missing = historical_features.isnull().sum()
historical_missing = historical_missing[historical_missing > 0].sort_values(ascending=False)

print("Missing values in historical features:")
display(historical_missing)
```

# Remove incomplete historical rows

```python
historical_features = historical_features.dropna().copy()

print("Historical features shape after dropping missing rows:")
print(historical_features.shape)
```

# Select final historical columns to merge

```python
historical_cols_to_keep = [
    "county_fips",
    
    "dem_share_2000",
    "dem_share_2004",
    "dem_share_2008",
    "dem_share_2012",
    "dem_share_2016",
    "dem_share_2020",
    
    "winner_2000",
    "winner_2004",
    "winner_2008",
    "winner_2012",
    "winner_2016",
    "winner_2020",
    
    "avg_dem_share_2000_2020",
    "dem_wins_count_2000_2020",
    "vote_volatility_2000_2020",
    "swing_2016_2020",
    "swing_2012_2020",
    "swing_2000_2020",
    
    "target_2024_party_winner",
    "target_2024_dem_vote_share"
]

historical_final = historical_features[historical_cols_to_keep].copy()

print(historical_final.shape)
display(historical_final.head())
```

# Merge historical features with model-ready ACS dataset

```python
final_historical_df = model_df.merge(
    historical_final,
    on="county_fips",
    how="inner"
)

print("Final historical ML dataset shape:")
print(final_historical_df.shape)

display(final_historical_df.head())
```

# Check target distribution for 2024

```python
print("2024 target class balance:")
print(final_historical_df["target_2024_party_winner"].value_counts())

print("\n2024 target class percentage:")
print(final_historical_df["target_2024_party_winner"].value_counts(normalize=True))

print("\n2024 Democratic vote share summary:")
display(final_historical_df["target_2024_dem_vote_share"].describe())
```

# Check missing values in final dataset

```python
missing_final = final_historical_df.isnull().sum()
missing_final = missing_final[missing_final > 0].sort_values(ascending=False)

print("Missing values in final historical dataset:")
display(missing_final)
```

# Remove Invalid Vote Shares

```python
final_historical_df = final_historical_df[
    final_historical_df["target_2024_dem_vote_share"] <= 1
].copy()

print("Dataset shape after removing invalid vote shares:")
print(final_historical_df.shape)

print("\nUpdated target vote share summary:")
display(
    final_historical_df["target_2024_dem_vote_share"].describe()
)
```

# Remove Remaining Missing Values

```python
final_historical_df = final_historical_df.dropna().copy()

print("Final cleaned dataset shape:")
print(final_historical_df.shape)

print("\nRemaining missing values:")
print(final_historical_df.isnull().sum().sum())
```

# Save final historical ML dataset

```python
os.makedirs("../data/processed", exist_ok=True)

final_historical_df.to_csv(output_path, index=False)

print("Historical model-ready dataset saved successfully:")
print(output_path)
```
