# Imports

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

# Evaluation Results

```python
# Full Historical Models

accuracy = 0.9690
precision = 0.8100
recall = 1.0000
f1 = 0.8950
roc_auc = 0.9968

accuracy_tree = 0.9788
precision_tree = 0.8696
recall_tree = 0.9877
f1_tree = 0.9249
roc_auc_tree = 0.9986

accuracy_rf = 0.9723
precision_rf = 0.8333
recall_rf = 0.9877
f1_rf = 0.9040
roc_auc_rf = 0.9972


# Demographics-Only Models

accuracy_demo = 0.8777
precision_demo = 0.5227
recall_demo = 0.8519
f1_demo = 0.6479
roc_auc_demo = 0.9413

accuracy_tree_demo = 0.8777
precision_tree_demo = 0.5224
recall_tree_demo = 0.8642
f1_tree_demo = 0.6512
roc_auc_tree_demo = 0.8784

accuracy_rf_demo = 0.9364
precision_rf_demo = 0.7561
recall_rf_demo = 0.7654
f1_rf_demo = 0.7607
roc_auc_rf_demo = 0.9636
```

# Create Overall Model Comparison Table

```python
model_results = pd.DataFrame({

    "Model": [

        "Logistic Regression",
        "Decision Tree",
        "Random Forest",

        "Demo-Only Logistic Regression",
        "Demo-Only Decision Tree",
        "Demo-Only Random Forest"
    ],

    "Accuracy": [

        accuracy,
        accuracy_tree,
        accuracy_rf,

        accuracy_demo,
        accuracy_tree_demo,
        accuracy_rf_demo
    ],

    "Precision": [

        precision,
        precision_tree,
        precision_rf,

        precision_demo,
        precision_tree_demo,
        precision_rf_demo
    ],

    "Recall": [

        recall,
        recall_tree,
        recall_rf,

        recall_demo,
        recall_tree_demo,
        recall_rf_demo
    ],

    "F1-Score": [

        f1,
        f1_tree,
        f1_rf,

        f1_demo,
        f1_tree_demo,
        f1_rf_demo
    ],

    "ROC-AUC": [

        roc_auc,
        roc_auc_tree,
        roc_auc_rf,

        roc_auc_demo,
        roc_auc_tree_demo,
        roc_auc_rf_demo
    ]
})

display(model_results)
```

# Accuracy Comparison Plot

```python
plt.figure(figsize=(12, 6))

sns.barplot(
    data=model_results,
    x="Accuracy",
    y="Model"
)

plt.title("Model Accuracy Comparison")

plt.xlim(0, 1)

plt.show()
```

# F1-Score Comparison Plot

```python
plt.figure(figsize=(12, 6))

sns.barplot(
    data=model_results,
    x="F1-Score",
    y="Model"
)

plt.title("Model F1-Score Comparison")

plt.xlim(0, 1)

plt.show()
```

# ROC-AUC Comparison Plot

```python
plt.figure(figsize=(12, 6))

sns.barplot(
    data=model_results,
    x="ROC-AUC",
    y="Model"
)

plt.title("Model ROC-AUC Comparison")

plt.xlim(0, 1)

plt.show()
```

# Historical vs Demo-Only Comparison

```python
historical_vs_demo = pd.DataFrame({

    "Model": [
        "Logistic Regression",
        "Decision Tree",
        "Random Forest"
    ],

    "Full Model Accuracy": [
        accuracy,
        accuracy_tree,
        accuracy_rf
    ],

    "Demo-Only Accuracy": [
        accuracy_demo,
        accuracy_tree_demo,
        accuracy_rf_demo
    ],

    "Accuracy Drop": [

        accuracy - accuracy_demo,

        accuracy_tree - accuracy_tree_demo,

        accuracy_rf - accuracy_rf_demo
    ],

    "Full Model F1": [
        f1,
        f1_tree,
        f1_rf
    ],

    "Demo-Only F1": [
        f1_demo,
        f1_tree_demo,
        f1_rf_demo
    ],

    "F1 Drop": [

        f1 - f1_demo,

        f1_tree - f1_tree_demo,

        f1_rf - f1_rf_demo
    ]
})

display(historical_vs_demo)
```

# Accuracy Drop Visualization

```python
plt.figure(figsize=(10, 5))

sns.barplot(
    data=historical_vs_demo,
    x="Accuracy Drop",
    y="Model"
)

plt.title("Accuracy Drop After Removing Historical Features")

plt.show()
```

# F1 Drop Visualization

```python
plt.figure(figsize=(10, 5))

sns.barplot(
    data=historical_vs_demo,
    x="F1 Drop",
    y="Model"
)

plt.title("F1-Score Drop After Removing Historical Features")

plt.show()
```

# Key Findings

### 1. Historical political behavior was the strongest predictor of county-level election outcomes.

### 2. Models using historical political features consistently outperformed demographics-only models.

### 3. Random Forest achieved the strongest demographics-only performance, indicating that nonlinear demographic interactions play an important role in political behavior.

### 4. Education-related variables emerged as the strongest demographic predictors of Democratic voting behavior.

### 5. Race composition variables such as white_pct, black_pct, and asian_pct significantly contributed to county political alignment.

### 6. Demographic and socioeconomic variables retained substantial independent predictive power even after removing historical political information.
